```python
import sys
from collections import deque

def longest_stable_window():
    # Efficient for large inputs up to N = 10^6
    input_data = sys.stdin.read().split()
    if not input_data:
        return

    n = int(input_data[0])
    arr = [int(x) for x in input_data[1:n + 1]]
    k = int(input_data[n + 1])

    max_deque = deque()  
    min_deque = deque()  

    left = 0
    max_len = 0
    best_start = 1  

    for right in range(n):
        val = arr[right]

        
        while max_deque and arr[max_deque[-1]] <= val:
            max_deque.pop()
        max_deque.append(right)

        
        while min_deque and arr[min_deque[-1]] >= val:
            min_deque.pop()
        min_deque.append(right)

        
        while arr[max_deque[0]] - arr[min_deque[0]] > k:
            if max_deque[0] == left:
                max_deque.popleft()
            if min_deque[0] == left:
                min_deque.popleft()
            left += 1

        
        current_len = right - left + 1
        
        
        if current_len > max_len:
            max_len = current_len
            best_start = left + 1  # Convert 0-based to 1-based index

    print(f"{max_len} {best_start}")

if __name__ == "__main__":
    longest_stable_window()
```
